## 1. Problem definition

**Câu hỏi:** Mỗi khu vực trong trung tâm Đà Nẵng thuộc “kiểu” gì nếu nhìn vào **cơ cấu ngành hàng ăn uống** của nó?

Ví dụ kỳ vọng: *khu cà phê – văn phòng*, *khu ăn sáng dân cư*, *khu ăn đêm / lẩu nướng*, *khu chợ truyền thống*.

**Ai dùng kết quả:** người muốn mở quán (chọn khu phù hợp mô hình), người đi ăn (biết khu nào mạnh món gì), nhóm nghiên cứu đô thị.

**Đầu ra mong muốn**
- Bản đồ chia ô, mỗi ô được gán một kiểu khu vực có tên dễ hiểu.
- Hồ sơ từng kiểu: cơ cấu ngành hàng, mức giá, đánh giá, từ khoá trong bình luận.
- Web tương tác để xem, khoanh vùng và tra cứu.

## 2. Analytic approach

Đây là bài toán **học không giám sát** (không có nhãn “kiểu khu vực” sẵn) → **phân cụm**.

| Thành phần | Lựa chọn | Lý do |
|---|---|---|
| Đơn vị không gian | **Lưới ô 250 m** | Không phụ thuộc ranh giới phường; lưới đều nhau, so sánh công bằng |
| Đặc trưng | Vector **tỷ lệ** các nhóm ngành hàng trong ô | Đúng định nghĩa “cơ cấu ngành hàng”, không phụ thuộc ô đông hay vắng |
| Giảm nhiễu | Gộp nhóm món, làm mượt không gian, biến đổi Hellinger, PCA | Mỗi ô chỉ có 5–40 quán nên tỷ lệ rất nhiễu |
| Mô hình | **K-Means**, **Hierarchical (Ward)**, **GMM** | So sánh 3 cách tiếp cận: tâm cụm, phân cấp, xác suất |
| Đánh giá | Silhouette, Davies–Bouldin, Calinski–Harabasz, độ ổn định (bootstrap ARI), mức đồng thuận giữa các mô hình, khả năng diễn giải | Không có nhãn thật nên cần nhiều góc nhìn |

Giá và bình luận **không đưa vào phân cụm** (đề tài phân loại theo cơ cấu ngành hàng); chúng dùng để **mô tả** từng kiểu sau khi phân cụm.

```python
SRC_XLSX = DATA / "cleaned" / "data_cleaned.xlsx"
USE = ["placeId", "title", "address", "neighborhood", "location/lat", "location/lng", "category", "categories", "totalScore", "reviewsCount",
       "permanentlyClosed", "temporarilyClosed", "stars", "publishedAtDate", "text", "textTranslated", "originalLanguage", "translatedLanguage",
       "priceMin_vnd", "priceMax_vnd", "likesCount"]
raw_reviews = pd.read_excel(SRC_XLSX, sheet_name="Data", usecols=USE, engine="openpyxl", keep_default_na=False, na_values=[""])

stores_raw = (raw_reviews.drop_duplicates("placeId")
              .rename(columns={"placeId": "storeId", "title": "storeName", "location/lat": "latitude", "location/lng": "longitude"})
              [["storeId", "storeName", "address", "neighborhood", "latitude", "longitude", "category", "categories", "totalScore", "reviewsCount",
                "permanentlyClosed", "temporarilyClosed"]].reset_index(drop=True))

# Giá mỗi quán: khoảng giá được nhiều người đánh giá chọn nhất, giữ nguyên khoảng gốc của Google; bằng phiếu thì lấy khoảng thấp hơn.
# Khoảng "Trên 1.000.000 ₫" không có cận trên. Quán không có ai khai giá thì để trống
prices_raw = (raw_reviews[raw_reviews.priceMin_vnd.notna()]
              .groupby(["placeId", "priceMin_vnd", "priceMax_vnd"], dropna=False).size().rename("n").reset_index()
              .sort_values(["placeId", "n", "priceMin_vnd"], ascending=[True, False, True]).drop_duplicates("placeId")
              .rename(columns={"placeId": "storeId", "priceMin_vnd": "price_min", "priceMax_vnd": "price_max"})[["storeId", "price_min", "price_max"]])

# Bình luận tiếng Việt: giữ nguyên bản tiếng Việt, tiếng khác dùng bản Google dịch sang tiếng Việt; không có bản tiếng Việt thì để trống
vi_text = raw_reviews.text.where(raw_reviews.originalLanguage == "vi")
reviews_raw = pd.DataFrame({
    "storeId": raw_reviews.placeId, "rating": raw_reviews.stars, "likes": raw_reviews.likesCount,
    "comment": raw_reviews.textTranslated.where(raw_reviews.translatedLanguage == "vi").fillna(vi_text),
    "created_at": (pd.to_datetime(raw_reviews.publishedAtDate) + pd.Timedelta(hours=7)).dt.date.astype(str),   # UTC sang giờ Việt Nam
})

print(f"Đánh giá : {raw_reviews.shape}  ← {SRC_XLSX.relative_to(ROOT)}")
print(f"Quán     : {stores_raw.shape}")
print(f"Giá      : {len(prices_raw)} quán có giá")
print(f"Bình luận: {reviews_raw.comment.notna().sum()} đánh giá có chữ tiếng Việt / {len(reviews_raw)}")
stores_raw.head(3)
```

## 6. Data preprocessing

1. Loại toạ độ ngoài Đà Nẵng, quán đã đóng cửa vĩnh viễn và quán trùng (cùng tên + cùng toạ độ).
2. Chuẩn hoá `category` → **nhóm món** (11 nhóm) → **nhóm ngành hàng lớn** (7 nhóm dùng để phân cụm); bỏ địa điểm không phải ăn uống.
3. Nối giá và bình luận theo `storeId`.
4. Chia lưới ô 250 m, giữ các ô có ít nhất 5 quán.
5. Tính vector cơ cấu ngành hàng có **làm mượt không gian**, biến đổi **Hellinger** (căn bậc hai tỷ lệ) và **PCA**.

```python
open_stores = stores_raw[~stores_raw.permanentlyClosed]
df = open_stores[open_stores.latitude.between(15.95, 16.15) & open_stores.longitude.between(108.10, 108.32)]
n0 = len(df)
df = df.drop_duplicates(["storeName", "latitude", "longitude"]).copy()
print(f"{len(stores_raw)} → {len(open_stores)} (bỏ quán đóng cửa vĩnh viễn) → {n0} (bỏ toạ độ sai) → {len(df)} (bỏ trùng)")
```

```python
# Nhóm món: so khớp category Google Maps theo thứ tự luật, nhãn đầu tiên (category chính) được ưu tiên
NON_FOOD = r"khách sạn|căn hộ|lưu trú|hiệu sách|chăm sóc móng|\bspa\b|sân chơi|giặt ủi|thể dục|studio|văn phòng|sửa chữa|thời trang|đồng phục|quà tặng|\bnến\b|dụng cụ|mỹ phẩm|thủ công"
GROUP_RULES = [
    ("ca_phe",    r"cà phê|phòng trà"),
    ("tra_sua",   r"trà trân châu|trà sữa|nước quả|giải khát"),
    ("banh_mi",   r"bánh mì|tiệm bánh|cửa hàng bánh|sandwich|bánh hấp"),
    ("cua_hang",  r"^cửa hàng$|tạp phẩm|tạp hóa|tiện lợi|thực phẩm|rau quả|trái cây|pho mát|bơ sữa|cửa hàng bia|cửa hàng rượu"),
    ("lau_nuong", r"lẩu|nướng|bít tết|yakiniku|món thịt|hải sản|izakaya|nhậu|\bbar\b|quán bia|quán rượu|karaoke|câu lạc bộ"),
    ("healthy",   r"\bchay\b|đậu hũ"),
    ("quoc_te",   r"hàn quốc|nhật bản|thái lan|trung quốc|quảng đông|tứ xuyên|nhà hàng ý|pháp|châu âu|nhà hàng anh|nhà hàng mỹ|mexico|ấn độ|thổ nhĩ kỳ|địa trung hải|phương tây|miến điện|sushi|pizza|ăn nhanh|gà rán|lai á|xuyên á|châu á|kebab|takoyaki|okonomiyaki|udon|ghana|đạo hồi"),
    ("bun_pho",   r"phở|\bmì\b|cháo|\bsúp\b|\bxúp\b"),
    ("an_vat",    r"\bkem\b|tráng miệng|ăn nhẹ|gian hàng rong|\bkẹo\b|sô cô la|sữa chua"),
    ("com",       r"\bcơm\b|thịt gà|quán ăn|đồ ăn|bữa sáng|bữa trưa|bữa nửa buổi|điểm tâm"),
]
GENERIC = {"nhà hàng", "nhà hàng gia đình", "khu ăn uống"}   # nhãn chung: xem thêm các nhãn phụ trước khi kết luận
DRINKS = {"ca_phe", "tra_sua"}
SUB_LABEL = {"ca_phe": "Cà phê", "tra_sua": "Trà sữa · nước ép", "bun_pho": "Bún · phở · mì", "com": "Cơm · quán ăn", "banh_mi": "Bánh mì · bánh",
             "an_vat": "Ăn vặt · chè", "lau_nuong": "Lẩu · nướng · quán nhậu", "quoc_te": "Món quốc tế", "healthy": "Healthy · chay",
             "nha_hang": "Nhà hàng món Việt", "cua_hang": "Cửa hàng · chợ", "khac": "Khác"}
# Nhóm ngành hàng lớn dùng để phân cụm (gộp các nhóm món gần nhau để giảm nhiễu)
MACRO = {"ca_phe": "do_uong", "tra_sua": "do_uong", "bun_pho": "an_sang", "banh_mi": "an_sang", "com": "com", "nha_hang": "com",
         "an_vat": "an_vat", "lau_nuong": "lau_nuong", "quoc_te": "quoc_te", "healthy": "quoc_te", "cua_hang": "cho_cua_hang"}
GROUPS = {"do_uong": ("Đồ uống", "Cà phê, trà sữa, nước ép"), "an_sang": ("Bún phở · bánh mì", "Món nước, bánh mì, bánh"),
          "com": ("Cơm · món Việt", "Cơm, quán ăn, nhà hàng món Việt"), "an_vat": ("Ăn vặt · chè", "Ăn vặt, kem, tráng miệng"),
          "lau_nuong": ("Lẩu · nướng", "Lẩu, nướng, hải sản, quán nhậu, bar"), "quoc_te": ("Quốc tế · healthy", "Pizza, gà rán, Hàn, Nhật, chay"),
          "cho_cua_hang": ("Chợ · cửa hàng", "Đồ khô, đặc sản, tạp hoá")}
G = list(GROUPS)
# Loại hình kinh doanh hiển thị trên web: gần cách người mở quán phân loại. Chỉ dùng để mô tả, không dùng để phân cụm
CATS = {"ca_phe": ("Cà phê", ["ca_phe"]), "banh_mi": ("Bánh mì", ["banh_mi"]), "tra_sua": ("Trà sữa · nước ép", ["tra_sua"]),
        "nha_hang": ("Nhà hàng", ["nha_hang", "lau_nuong", "quoc_te", "healthy"]), "cua_hang": ("Cửa hàng · chợ", ["cua_hang"]),
        "quan_an": ("Quán cơm · bún phở", ["com", "bun_pho"]), "an_vat": ("Ăn vặt · chè", ["an_vat"])}
CAT_OF = {s: c for c, (_, subs) in CATS.items() for s in subs}
C = list(CATS)

def match_group(label):
    for g, pattern in GROUP_RULES:
        if re.search(pattern, label):
            return g
    return None

def food_group(name, category, categories):
    labels = [x.strip().lower() for x in str(categories).split("|")] if pd.notna(categories) else []
    if not labels:   # Google không gán nhãn: chỉ nhận ra quán cà phê qua tên
        return "ca_phe" if re.search(r"cafe|cà phê|coffee", str(name).lower()) else "khac"
    main = labels[0]
    if re.search(NON_FOOD, main):   # ví dụ khách sạn có nhà hàng: chỉ tính nếu nhãn phụ là ăn uống
        return next((g for g in map(match_group, labels[1:]) if g), "khac")
    if main in GENERIC:
        return next((g for g in map(match_group, labels[1:]) if g and g not in DRINKS), "nha_hang")
    return match_group(main) or ("nha_hang" if main.startswith("nhà hàng") else "khac")

df["sub"] = [food_group(n, c, cs) for n, c, cs in zip(df.storeName, df.category, df.categories)]
df["g"] = df["sub"].map(MACRO)
summary = df.groupby(["g", "sub"], dropna=False).size().rename("số quán").reset_index()
print(f"Bỏ {df.g.isna().sum()} quán nhóm 'Khác' (không phải địa điểm ăn uống): {df[df.g.isna()].category.value_counts().to_dict()}")
df = df[df.g.notna()].copy()
df["biz"] = df["sub"].map(CAT_OF)
summary
```

```python
# Giá: giữ nguyên khoảng giá gốc của từng quán; điểm và số đánh giá: tổng trên Google Maps
def common_band(s):
    # Khoảng giá nhiều quán có nhất trong nhóm, bằng nhau thì lấy khoảng thấp hơn; không quán nào có giá thì để trống
    v = s[s.price_min.notna()]
    if v.empty:
        return None
    a, b = v.groupby(["price_min", "price_max"], dropna=False).size().rename("n").reset_index().sort_values(["n", "price_min"], ascending=[False, True]).iloc[0][["price_min", "price_max"]]
    return [int(a), None if pd.isna(b) else int(b)]

rev = reviews_raw.dropna(subset=["comment"]).sort_values("created_at", ascending=False)
df = df.merge(prices_raw, on="storeId", how="left")
df["rating"], df["n_reviews"] = df.totalScore, df.reviewsCount.astype(int)
print(f"Có giá: {df.price_min.notna().mean():.0%} · có bình luận: {df.storeId.isin(rev.storeId).mean():.0%}")
df[["storeName", "sub", "g", "price_min", "price_max", "rating", "n_reviews"]].head()
```

```python
# Lưới ô 250 m: chiếu toạ độ ra mét quanh vĩ độ trung bình (sai số không đáng kể ở phạm vi vài km)
CELL = 250        # cạnh ô (m)
MIN_STORES = 5    # ô có ít hơn số quán này bị bỏ vì tỷ lệ quá nhiễu
BW = 250          # bán kính làm mượt (m)
LNG0, LAT0 = 108.15, 15.95
KX, KY = 111_320 * np.cos(np.radians(df.latitude.mean())), 110_574

df["x"], df["y"] = (df.longitude - LNG0) * KX, (df.latitude - LAT0) * KY
df["cx"], df["cy"] = (df.x // CELL).astype(int), (df.y // CELL).astype(int)
df["cell"] = df.cx.astype(str) + "_" + df.cy.astype(str)

cnt = df.groupby("cell").size()
cells = pd.DataFrame({"cell": cnt[cnt >= MIN_STORES].index})
cells[["cx", "cy"]] = cells.cell.str.split("_", expand=True).astype(int)
print(f"{df.cell.nunique()} ô có quán → giữ {len(cells)} ô (≥ {MIN_STORES} quán), "
      f"chứa {int(cnt[cnt >= MIN_STORES].sum())}/{len(df)} quán")

fig, ax = plt.subplots(figsize=(6, 6))
ax.hist(cnt, bins=30, color="#4A90B5"); ax.axvline(MIN_STORES, color="#F28C5B", ls="--")
ax.set_title("Số quán mỗi ô 250 m"); plt.show()
```

```python
# Cơ cấu thô (đếm trong ô) và cơ cấu làm mượt: mỗi quán đóng góp cho tâm ô với trọng số Gauss theo khoảng cách
raw = pd.crosstab(df.cell, df.g).reindex(index=cells.cell, columns=G, fill_value=0)
centers = np.c_[(cells.cx + .5) * CELL, (cells.cy + .5) * CELL]
onehot = pd.get_dummies(df.g).reindex(columns=G, fill_value=0).values.astype(float)
dist2 = ((centers[:, None, :] - df[["x", "y"]].values[None, :, :]) ** 2).sum(-1)
W = np.exp(-dist2 / (2 * BW ** 2))
comp = W @ onehot
comp = comp / comp.sum(1, keepdims=True)

X_hel = np.sqrt(comp)                      # Hellinger: phù hợp hơn cho dữ liệu tỷ lệ khi dùng khoảng cách Euclid
pca = PCA(n_components=3, random_state=SEED).fit(X_hel)
Z = pca.transform(X_hel)
print("Tỷ lệ phương sai giải thích bởi 3 thành phần PCA:", np.round(pca.explained_variance_ratio_, 3),
      "→ tổng", round(pca.explained_variance_ratio_.sum(), 3))

features = pd.DataFrame(comp, columns=[f"p_{g}" for g in G]).assign(cell=cells.cell.values, n_stores=raw.sum(1).values)
features.head()
```

**Vì sao cần làm mượt và gộp nhóm?** Ở bản nháp, phân cụm thẳng tỷ lệ thô của 11 nhóm món cho silhouette chỉ khoảng 0.1: mỗi ô quá ít quán so với số nhóm. Gộp còn 7 nhóm và làm mượt bán kính 250 m (mỗi ô “mượn” thông tin từ quán lân cận) giúp cấu trúc rõ hơn. Bán kính được giữ vừa phải vì làm mượt quá tay sẽ khiến các ô kề nhau giống nhau một cách nhân tạo và làm điểm số đẹp giả.

## 7. Model

Chạy 3 mô hình với số cụm k = 3…7 trên không gian PCA:

- **K-Means**: chia theo tâm cụm, nhanh, dễ diễn giải.
- **Hierarchical (Ward)**: gộp dần các ô giống nhau, xem được cây phân cấp.
- **GMM**: mỗi cụm là một phân phối Gauss, cho xác suất thuộc cụm.

Độ ổn định của K-Means được đo bằng bootstrap: lấy mẫu lại các ô 30 lần, huấn luyện lại, rồi so nhãn với mô hình gốc (ARI).

```python
def stability(Z, k, n_boot=30):
    base = KMeans(k, n_init=20, random_state=SEED).fit_predict(Z)
    rng = np.random.default_rng(SEED)
    scores = []
    for b in range(n_boot):
        idx = rng.choice(len(Z), len(Z), replace=True)
        m = KMeans(k, n_init=10, random_state=b).fit(Z[idx])
        scores.append(adjusted_rand_score(base, m.predict(Z)))
    return float(np.mean(scores))

MODELS = {
    "K-Means": lambda k: KMeans(k, n_init=50, random_state=SEED).fit_predict(Z),
    "Hierarchical": lambda k: AgglomerativeClustering(k, linkage="ward").fit_predict(Z),
    "GMM": lambda k: GaussianMixture(k, covariance_type="full", n_init=10, random_state=SEED).fit(Z).predict(Z),
}
rows, labels = [], {}
for k in range(3, 8):
    for name, fit in MODELS.items():
        lab = fit(k); labels[(name, k)] = lab
        rows.append({"model": name, "k": k, "silhouette": silhouette_score(Z, lab),
                     "davies_bouldin": davies_bouldin_score(Z, lab), "calinski_harabasz": calinski_harabasz_score(Z, lab),
                     "stability": stability(Z, k) if name == "K-Means" else np.nan})
scores = pd.DataFrame(rows)
scores.pivot(index="k", columns="model", values="silhouette").round(3)
```

```python
fig, ax = plt.subplots(1, 3, figsize=(14, 3.6))
for name, g in scores.groupby("model"):
    ax[0].plot(g.k, g.silhouette, marker="o", label=name)
    ax[1].plot(g.k, g.davies_bouldin, marker="o", label=name)
km = scores[scores.model == "K-Means"]
ax[2].plot(km.k, km.stability, marker="o", color="#4A90B5")
ax[0].set_title("Silhouette (cao hơn tốt hơn)"); ax[1].set_title("Davies–Bouldin (thấp hơn tốt hơn)")
ax[2].set_title("Độ ổn định K-Means (bootstrap ARI)"); ax[0].legend()
for a in ax: a.set_xlabel("k")
plt.tight_layout(); plt.show()
```

```python
# Đặt tên theo đúng dữ liệu: 1–2 nhóm ngành hàng vượt mức chung nhiều nhất, để tên giải thích được bằng số liệu và tự đúng khi chạy lại
NAME_OF = {"do_uong": "cà phê", "an_sang": "bún phở · bánh mì", "com": "cơm · món Việt", "an_vat": "ăn vặt",
           "lau_nuong": "lẩu nướng · nhậu", "quoc_te": "món quốc tế", "cho_cua_hang": "chợ · cửa hàng"}
MIN_LIFT = 1.15   # lift thấp hơn nghĩa là cơ cấu gần mức chung: gọi là khu hỗn hợp thay vì gán tên theo một nhóm
MIN_SHARE = .05   # nhóm chiếm dưới 5% cụm thì lift cao chỉ là nhiễu của nhóm rất nhỏ, không dùng để đặt tên hay mô tả

def top_groups(c):
    top = lift.loc[c].sort_values(ascending=False)
    return top[(top >= MIN_LIFT) & (prof.loc[c, top.index] >= MIN_SHARE)].head(2)

type_name = {}
for c in range(K):
    name = "Khu " + " – ".join(NAME_OF[g] for g in top_groups(c).index) if len(top_groups(c)) else "Khu hỗn hợp"
    n_same = sum(v.startswith(name) for v in type_name.values())
    type_name[c] = f"{name} {n_same + 1}" if n_same else name

def why_name(c):
    # Số liệu giải thích tên, web tự trình bày: các nhóm làm nên tên và nhóm thấp hơn mức chung rõ nhất
    top = [{"g": g, "share": round(float(prof.loc[c, g]), 4), "lift": round(float(v), 2)} for g, v in top_groups(c).items()]
    low = lift.loc[c, city_share[city_share >= MIN_SHARE].index].idxmin()
    below = {"g": low, "share": round(float(prof.loc[c, low]), 4), "lift": round(float(lift.loc[c, low]), 2)} if lift.loc[c, low] < 1 / MIN_LIFT else None
    return {"top": top, "low": below}

def describe(c):
    top = lift.loc[c].sort_values(ascending=False)
    top = top[(top > 1.05) & (prof.loc[c, top.index] >= MIN_SHARE)].head(2)
    if top.empty:
        return "Cơ cấu gần với mức chung của thành phố"
    return "Nhiều " + " và ".join(f"{GROUPS[g][0]} (×{v:.1f})" for g, v in top.items()) + " hơn mức chung"

types = pd.DataFrame({"cụm": range(K), "tên": [type_name[c] for c in range(K)], "mô tả": [describe(c) for c in range(K)],
                      "số ô": [int((label == c).sum()) for c in range(K)],
                      "số quán": [int(raw[label == c].values.sum()) for c in range(K)]})
types
```

```python
# Mô tả thêm bằng giá và bình luận (không dùng để phân cụm)
cell_type = dict(zip(cells.cell, label))
df["type"] = df.cell.map(cell_type)

def phrases(texts):
    # Tách bình luận thành các cụm ngắn theo dấu câu, giữ các cụm 2–4 từ
    c = Counter()
    for t in texts:
        for part in re.split(r"[,.!?;\n]", t.lower()):
            w = part.strip().split()
            if 2 <= len(w) <= 4:
                c[" ".join(w)] += 1
    return c

ALL_PHRASES = phrases(reviews_raw.comment.dropna())
N_ALL = sum(ALL_PHRASES.values())

def keywords(texts, top=6, min_count=3, min_lift=1.2):
    # Từ khoá ĐẶC TRƯNG: cụm từ xuất hiện trong nhóm nhiều hơn hẳn mức chung (tỷ lệ × log tần suất),
    # để tránh các câu chung chung kiểu "sẽ quay lại" xuất hiện ở mọi nơi. Bỏ cụm có chữ số (giá tiền, số nhà)
    c = phrases(texts.dropna())
    n = sum(c.values()) or 1
    lift_of = lambda w, k: (k / n) / (ALL_PHRASES[w] / N_ALL)
    score = {w: lift_of(w, k) * np.log1p(k) for w, k in c.items() if k >= min_count and lift_of(w, k) >= min_lift and not re.search(r"\d", w)}
    return [w for w, _ in sorted(score.items(), key=lambda x: -x[1])[:top]]

rev_with_type = reviews_raw.merge(df[["storeId", "type", "cell"]], on="storeId")
prof_extra = df[df.type.notna()].groupby("type").agg(danh_gia=("rating", "mean"),
                                                     binh_luan=("n_reviews", "sum"))
prof_extra["từ khoá"] = [", ".join(keywords(rev_with_type[rev_with_type.type == c].comment, 4)) for c in prof_extra.index]
type_ids = prof_extra.index
prof_extra.index = [type_name[int(c)] for c in type_ids]
prof_extra["khoảng giá phổ biến"] = [common_band(df[df.type == type_id]) for type_id in type_ids]
prof_extra.round(2)
```

- k = 7 được chọn để bản đồ chi tiết hơn, không phải vì chỉ số: k = 4 có silhouette (0,38) và độ ổn định (0,73) cao hơn k = 7 (0,63). Ba mô hình đồng thuận ở mức trung bình (ARI ~0.57–0.61), nên ranh giới chi tiết có phụ thuộc vào thuật toán.

**Nạp dữ liệu mới:** cập nhật `data/cleaned/data_cleaned.xlsx` rồi chạy lại notebook, file JSON được ghi đè, tải lại trang web. Web cũng có nút *Nạp dữ liệu* để mở một file JSON khác.

```python
# Màu kiểu khu vực (bản sáng). Web dùng bảng màu riêng cho chế độ sáng/tối, đã kiểm tra khả năng phân biệt
# (kể cả người mù màu) trên các cặp kiểu khu vực giáp nhau trên bản đồ.
PALETTE = ["#2a78d6", "#1baf7a", "#eda100", "#e34948", "#6250d6", "#008300", "#e87ba4", "#eb6834"]
to_ll = lambda x, y: [round(LAT0 + y / KY, 6), round(LNG0 + x / KX, 6)]
row = scores[(scores.model == METHOD) & (scores.k == K)].iloc[0]

# VÙNG: gộp các ô kề nhau (4 hướng) cùng kiểu khu vực thành một vùng liền mạch, đánh số theo diện tích
pos = {(r.cx, r.cy): i for i, r in cells.iterrows()}
zone_of, comps = {}, []
for start in pos:
    if pos[start] in zone_of:
        continue
    t, stack, comp = label[pos[start]], [start], []
    zone_of[pos[start]] = -1
    while stack:
        x, y = stack.pop(); comp.append(pos[(x, y)])
        for q in ((x + 1, y), (x - 1, y), (x, y + 1), (x, y - 1)):
            if q in pos and pos[q] not in zone_of and label[pos[q]] == t:
                zone_of[pos[q]] = -1; stack.append(q)
    comps.append((int(t), comp))
comps.sort(key=lambda c: -len(c[1]))
for z, (_, comp) in enumerate(comps):
    for i in comp:
        zone_of[i] = z
cells["zone"] = [zone_of[i] for i in range(len(cells))]
df["zone"] = df.cell.map(dict(zip(cells.cell, cells.zone)))
rev_with_type = rev_with_type.merge(df[["storeId", "zone", "g", "biz"]], on="storeId")

# Xu hướng: số bình luận mỗi ngày theo loại hình kinh doanh, 90 ngày gần nhất của dữ liệu
TREND_DAYS = 90
rev_dates = pd.to_datetime(rev_with_type.created_at.where(rev_with_type.comment.notna()), errors="coerce")
trend_end = rev_dates.max().normalize() if rev_dates.notna().any() else None
def trend(mask):
    if trend_end is None:
        return None
    r = rev_with_type[mask & rev_dates.notna()]
    day = (trend_end - pd.to_datetime(r.created_at).dt.normalize()).dt.days
    keep = (day >= 0) & (day < TREND_DAYS)
    r, day = r[keep], day[keep]
    out = {}
    for c in C:
        v = np.zeros(TREND_DAYS, int)
        np.add.at(v, TREND_DAYS - 1 - day[r.biz == c].values, 1)
        out[c] = v.tolist()
    return out

# HÌNH VÙNG: ghép các ô, thu viền 5 m, bo góc rồi cắt theo bờ sông để viền chạy dọc bờ và diện tích chỉ tính phần đất
INSET_M, ROUND_M = 5, 65
to_m = lambda lng, lat, z=None: ((lng - LNG0) * KX, (lat - LAT0) * KY)
to_deg = lambda x, y, z=None: (LNG0 + x / KX, LAT0 + y / KY)
water = transform(to_m, unary_union([shape(f["geometry"]) for f in json.loads((DATA / "geo" / "water.geojson").read_text(encoding="utf-8"))["features"]]))

def zone_shape(zc):
    cells_m = unary_union([box(r.cx * CELL, r.cy * CELL, (r.cx + 1) * CELL, (r.cy + 1) * CELL) for r in zc.itertuples()])
    soft = (cells_m.buffer(-INSET_M, join_style="mitre").buffer(-ROUND_M, join_style="mitre").buffer(ROUND_M)
            .buffer(ROUND_M).buffer(-ROUND_M))
    land = soft.difference(water).buffer(-6).buffer(6)   # bỏ các dải đất rất mảnh do cắt theo bờ
    geom = transform(to_deg, land)
    polys = list(geom.geoms) if geom.geom_type == "MultiPolygon" else [geom]
    coords = [[[[round(x, 6), round(y, 6)] for x, y in ring] for ring in mapping(p)["coordinates"]] for p in polys]
    return {"type": "MultiPolygon", "coordinates": coords}, cells_m.difference(water).area / 1e6

def on_land(r):
    return not water.contains(box(r.cx * CELL, r.cy * CELL, (r.cx + 1) * CELL, (r.cy + 1) * CELL).centroid)

short = lambda t: t if len(t) <= 220 else t[:217].rstrip() + "…"
POS_STARS = 4   # đánh giá từ 4 sao trở lên được xem là tích cực

# Bình luận hiển thị: 5 sao, ngắn gọn, có từ khen và không có từ chê, vì người chấm cao vẫn có thể viết lời chê
POS_WORDS = r"ngon|tuyệt|thích|đẹp|thân thiện|nhiệt tình|hài lòng|xuất sắc|tận tình|đáng thử|sẽ quay lại|rất ổn|chu đáo"
NEG_WORDS = r"\b(?:không|chưa|chẳng|nhưng|tuy|hơi|tóc|dở|tệ|kém|thất vọng|hạn chế|đắt|chậm|bẩn|ít|khó|mỗi tội|trừ|đừng|tạm|tốt hơn|thay đổi giá|cô gái)\b|[\u3400-\u9fff]"

def positive_comments(r, top=5):
    # Nhiều lượt thích hơn hiện trước, mỗi quán một bình luận
    t = r[(r.rating == 5) & r.comment.str.len().between(30, 220)]
    low = t.comment.str.lower()
    c = t[low.str.contains(POS_WORDS) & ~low.str.contains(NEG_WORDS)].sort_values(["likes", "created_at"], ascending=False).drop_duplicates("storeId").head(top)
    return [[a, int(b), short(x)] for a, b, x in zip(c.storeId, c.rating, c.comment)]

zone_rows = []
for z, (t, comp) in enumerate(comps):
    zc = cells.iloc[comp]
    s = df[df.zone == z]
    cnt_z = raw.loc[zc.cell].sum()
    geom, land_km2 = zone_shape(zc)
    mid = np.c_[(zc.cx + .5) * CELL, (zc.cy + .5) * CELL]
    order = ((mid - mid.mean(0)) ** 2).sum(1).argsort()
    anchor = next((zc.iloc[int(i)] for i in order if on_land(zc.iloc[int(i)])), zc.iloc[int(order[0])])   # ô gần tâm vùng nhất: đặt nhãn và biểu đồ
    zone_rows.append({
        "id": z, "name": f"Vùng {z + 1:02d}", "t": t, "cells": zc.cell.tolist(), "n_priced": int(s.price_min.notna().sum()),
        "geom": geom, "km2": round(land_km2, 4), "n": int(len(s)),
        "counts": {g: int(cnt_z[g]) for g in G}, "cats": {c: int((s.biz == c).sum()) for c in C},
        "anchor": to_ll((anchor.cx + .5) * CELL, (anchor.cy + .5) * CELL),
        "price": common_band(s),
        "rating": None if s.rating.isna().all() else round(float(s.rating.mean()), 2),
        "reviews": int(s.n_reviews.sum()), "kw": keywords(rev_with_type[rev_with_type.zone == z].comment, 6),
        "n_rated": int((rev_with_type.zone == z).sum()), "pos_rate": round(float((rev_with_type[rev_with_type.zone == z].rating >= POS_STARS).mean()), 4),
        "pos_cm": positive_comments(rev_with_type[rev_with_type.zone == z]),
        "trend": trend(rev_with_type.zone == z),
    })
print(f"{len(zone_rows)} vùng liền mạch; {sum(len(c) >= 3 for _, c in comps)} vùng có từ 3 ô trở lên")

cell_rows = []
for i, r in cells.iterrows():
    s = df[df.cell == r.cell]
    rv = rev_with_type[rev_with_type.cell == r.cell]
    cell_rows.append({
        "id": r.cell, "t": int(label[i]), "z": int(r.zone),
        "b": [to_ll(r.cx * CELL, r.cy * CELL), to_ll((r.cx + 1) * CELL, (r.cy + 1) * CELL)],
        "n": int(len(s)), "counts": {g: int(raw.loc[r.cell, g]) for g in G if raw.loc[r.cell, g]},
        "cats": {c: int(k) for c, k in s.biz.value_counts().items()},
        "price": common_band(s),
        "rating": None if s.rating.isna().all() else round(float(np.average(s.rating.dropna(), weights=s.n_reviews[s.rating.notna()])), 2),
        "reviews": int(s.n_reviews.sum()), "kw": keywords(rv.comment, 5, min_count=2),
    })

type_rows = []
for c in range(K):
    s = df[df.type == c]
    type_rows.append({
        "id": c, "name": type_name[c], "desc": describe(c), "why": why_name(c), "color": PALETTE[c % len(PALETTE)],
        "cells": int((label == c).sum()), "stores": int(len(s)),
        "share": {g: round(float(prof.loc[c, g]), 4) for g in G}, "lift": {g: round(float(lift.loc[c, g]), 2) for g in G},
        "price": common_band(s),
        "rating": None if s.rating.isna().all() else round(float(s.rating.mean()), 2),
        "kw": keywords(rev_with_type[rev_with_type.type == c].comment, 6),
    })

# 3 bình luận mới nhất mỗi quán, cắt ngắn để hiện vừa popup trên web
top_comments = {k: [[int(a), short(b)] for a, b in zip(x.rating, x.comment)] for k, x in rev.groupby("storeId").head(3).groupby("storeId")}
store_rows = [{
    "id": s.storeId, "n": s.storeName, "la": round(s.latitude, 6), "ln": round(s.longitude, 6), "g": s.g, "k": s.biz, "sub": SUB_LABEL[s["sub"]],
    "pmin": None if pd.isna(s.price_min) else int(s.price_min), "pmax": None if pd.isna(s.price_max) else int(s.price_max),
    "r": None if pd.isna(s.rating) else round(float(s.rating), 1), "rc": int(s.n_reviews),
    "c": s.cell if s.cell in cell_type else None, "cm": top_comments.get(s.storeId, []),
} for _, s in df.iterrows()]

out = {
    "meta": {
        "title": "Kiểu khu vực ẩm thực Đà Nẵng", "generated_at": datetime.now().isoformat(timespec="seconds"),
        "source": {"stores": SRC_XLSX.name, "prices": "crawl", "reviews": "crawl"},
        "cell_m": CELL, "min_stores": MIN_STORES, "smooth_m": BW, "method": METHOD, "k": K,
        "metrics": {"silhouette": round(float(row.silhouette), 3), "davies_bouldin": round(float(row.davies_bouldin), 3),
                    "stability": round(float(row.stability), 3),
                    "ari_hc": float(agree.loc[METHOD, "Hierarchical"]), "ari_gmm": float(agree.loc[METHOD, "GMM"])},
        "n_stores": len(df), "n_cells": len(cells), "n_zones": len(zone_rows),
        "trend_end": None if trend_end is None else trend_end.date().isoformat(), "trend_days": TREND_DAYS,
    },
    "zones": zone_rows,
    "groups": [{"id": g, "label": GROUPS[g][0], "detail": GROUPS[g][1], "city": round(float(city_share[g]), 4)} for g in G],
    "cats": [{"id": c, "label": CATS[c][0], "detail": ", ".join(SUB_LABEL[x] for x in CATS[c][1])} for c in C],
    "types": type_rows, "cells": cell_rows, "stores": store_rows,
}
path = DATA / "web" / "areas.json"
path.write_text(json.dumps(out, ensure_ascii=False, separators=(",", ":")), encoding="utf-8")
print(f"Đã ghi {path.relative_to(ROOT)} ({path.stat().st_size / 1024:.0f} KB): "
      f"{len(type_rows)} kiểu, {len(cell_rows)} ô, {len(store_rows)} quán")
```
